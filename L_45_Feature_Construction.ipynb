```python
import numpy as np
import pandas as pd

from sklearn.model_selection import cross_val_score
from sklearn.linear_model import LogisticRegression

import seaborn as sns
```

```python
df = pd.read_csv('train.csv')[['Age','Pclass','SibSp','Parch','Survived']]
```

```python
df.head()
```

```python
df.dropna(inplace=True)
```

```python
df.head()
```

```python
X = df.iloc[:,0:4]
y = df.iloc[:,-1]
```

* iloc: This is a method in pandas used
for integer-location based indexing. It allows you to select rows and columns by their numerical position (from 0 to N-1), rather than by their labels or conditions.

* df.iloc[:,0:4]:

The first part, [:] before the comma, means 'select all rows'.
The second part, 0:4 after the comma, means 'select columns from index 0 up to, but not including, index 4'. So, it selects columns at integer positions 0, 1, 2, and 3.
In your code, X = df.iloc[:,0:4] assigns the first four columns of all rows from df to the DataFrame X.

* df.iloc[:,-1]:

* Again, [:] before the comma means 'select all rows'.
The -1 after the comma means 'select the last column'. Negative indices count from the end, so -1 refers to the very last column.
In your code, y = df.iloc[:,-1] assigns the last column of all rows from df to the Series y.

```python
X.head()
```

```python
np.mean(cross_val_score(LogisticRegression(),X,y,scoring='accuracy',cv=20))
```

**Applying Feature Construction**

```python
X['Family_size'] = X['SibSp'] + X['Parch'] + 1
```

```python
X.head()
```

```python
def myfunc(num):
    if num == 1:
        #alone
        return 0
    elif num >1 and num <=4:
        # small family
        return 1
    else:
        # large family
        return 2
```

```python
myfunc(4)
```

```python
X['Family_type'] = X['Family_size'].apply(myfunc)
```

```python
X.head()
```

```python
X.drop(columns=['SibSp','Parch','Family_size'],inplace=True)
```

```python
X.head()
```

```python
np.mean(cross_val_score(LogisticRegression(),X,y,scoring='accuracy',cv=20))
```
